```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
```

```python
cam = cv2.VideoCapture(0)
ok, frame = cam.read()
if ok:
    cv2.imwrite("captured_frame.jpg", frame)
cam.release()
```

```python
saved = cv2.imread("captured_frame.jpg")
```

```python
plt.imshow(saved[:, :, ::-1])
plt.title("Captured Frame")
plt.axis("off")
plt.show()
```

Show the video

```python
cam = cv2.VideoCapture(0)
for n in range(30):
    ok, frame = cam.read()
    if not ok:
        break
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    clear_output(wait=True)
    plt.imshow(rgb)
    plt.axis("off")
    plt.show()
    time.sleep(0.05)
cam.release()
```

Show it in a smaller window

```python
cam = cv2.VideoCapture(0)
for n in range(30):
    ok, frame = cam.read()
    if not ok:
        break
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    clear_output(wait=True)
    plt.figure(figsize=(5, 3))
    plt.imshow(rgb)
    plt.axis("off")
    plt.show()
    time.sleep(0.05)
cam.release()
```

Rotate the video and show it

```python
cam = cv2.VideoCapture(0)
for n in range(30):
    ok, frame = cam.read()
    if not ok:
        break
    turned = cv2.rotate(frame, cv2.ROTATE_90_CLOCKWISE)
    rgb = cv2.cvtColor(turned, cv2.COLOR_BGR2RGB)
    clear_output(wait=True)
    plt.imshow(rgb)
    plt.axis("off")
    plt.show()
    time.sleep(0.05)
cam.release()
```
